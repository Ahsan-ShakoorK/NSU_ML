```python
df = pd.read_csv('study_score.csv')
df
```

```python
from sklearn.metrics import accuracy_score

from sklearn.tree import DecisionTreeRegressor

class RegressionTree:
    def __init__(self, max_depth=3):
        self.model = DecisionTreeRegressor(max_depth=max_depth, random_state=0)

    def fit(self, X, y):
        self.model.fit(X, y)
        return self

    def predict(self, X):
        return self.model.predict(X)
```

## Decision tree classifier

```python
# TODO: Write a classification tree class that can build tree and make prediction
from sklearn.tree import DecisionTreeClassifier

class ClassificationTree:
    def __init__(self, max_depth=3):
        self.model = DecisionTreeClassifier(max_depth=max_depth, random_state=0)

    def fit(self, X, y):
        self.model.fit(X, y)
        return self

    def predict(self, X):
        return self.model.predict(X)
```

```python
# Предсказания на обучающих данных
from sklearn.metrics import mean_squared_error, r2_score

X_reg = df.drop(columns=['Score'])
y_reg = df['Score']

tree = RegressionTree(max_depth=4).fit(X_reg, y_reg)
y_pred_reg = tree.predict(X_reg)
print("----Regression Results----")
print(f"Predicted Scores: {y_pred_reg}")
print(f"Mean Squared Error: {mean_squared_error(y_reg, y_pred_reg):.4f}")
print(f"R2 Score: {r2_score(y_reg, y_pred_reg):.4f}")
```

```python
# Метрики качества
mse = mean_squared_error(y_reg, y_pred_reg)
print("MSE: ", round(mse, 4))
print("RMSE:", round(np.sqrt(mse), 4))
print("R2:  ", round(r2_score(y_reg, y_pred_reg), 4))
```

```python

X_clf = df.drop(columns=['Score', 'Pass'])
y_clf = df['Pass']


clf = ClassificationTree(max_depth=4).fit(X_clf, y_clf)
y_pred_clf = clf.predict(X_clf)

print("--- Classification Results ---")
print(f"True Labels:      {y_clf.values}")
print(f"Predicted Labels: {y_pred_clf}")
print(f"Best Accuracy: {accuracy_score(y_clf, y_pred_clf) * 100:.2f}%")
```
